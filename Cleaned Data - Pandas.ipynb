```python
import pandas as pd
```

```python
df = pd.read_csv("D:/Airlines_Data_Analytics_Uncleaned.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.replace(["NULL", "N/A", "null", "n/a"], pd.NA, inplace=True)
```

```python
df["Aircraft_Type"] = df["Aircraft_Type"].fillna("Unknown")

df["Internal_Note"] = df["Internal_Note"].fillna("Not Available")
```

```python
df["Passenger_Count"] = pd.to_numeric(
    df["Passenger_Count"], errors="coerce")

df["Passenger_Count"] = df["Passenger_Count"].fillna(
    df["Passenger_Count"].median()).astype(int)
```

```python
df["Ticket_Price_INR"] = (
    df["Ticket_Price_INR"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace("INR", "", regex=False)
    .str.strip())

df["Ticket_Price_INR"] = pd.to_numeric(
    df["Ticket_Price_INR"], errors="coerce")

df["Ticket_Price_INR"] = df["Ticket_Price_INR"].fillna(
    df["Ticket_Price_INR"].median())
```

```python
df.drop_duplicates(inplace=True)
```

```python
df["Airline"] = df["Airline"].str.strip()

df["Origin"] = df["Origin"].str.strip()

df["Destination"] = df["Destination"].str.strip()
```

```python
# Airline
df["Airline"] = df["Airline"].replace({
    "INDIGO": "IndiGo",
    "AIR INDIA": "Air India",
    "spicejet": "SpiceJet"})

#Flight Status
df["Flight_Status"] = df["Flight_Status"].replace({
    "on-time": "On Time",
    "ON TIME": "On Time",
    "DELAYED": "Delayed",
    "cancelled": "Cancelled"})

#Booking Channel
df["Booking_Channel"] = df["Booking_Channel"].replace({
    "website": "Website",
    "MOBILE APP": "Mobile App"})

#Cabin Class
df["Cabin_Class"] = df["Cabin_Class"].replace({
    "economy": "Economy",
    "BUSINESS": "Business"})
```

```python
df["Flight_Date"] = pd.to_datetime(
    df["Flight_Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce")
```

```python
df["Flight_Date"] = df["Flight_Date"].fillna(
    df["Flight_Date"].median())
```

```python
df.drop(columns=["Internal_Note"], inplace=True)
```
